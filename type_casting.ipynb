2. explicit
- int()
- float()
- string()
- bool()

```python
a = 2.99
b = "2"
c = True # True=>1, False=>0

int(a), int(b), int(c)
```

```python
d = round(a)
d
```

```python
a = 2
b = "4.32"
c = True # True=>1.0, False=>0.0

float(a), float(b), float(c)
```

```python
a = 2
b = 2.31
c = True #True=>'True', False=>'False'
str(a), str(b), str(c)
```

```python
a = 3 # 0->False, otherwise->True
b = -31.54 # 0->False, otherwise->True
c = "x" # ""->False, otherwise->True

bool(a), bool(b), bool(c)
```
